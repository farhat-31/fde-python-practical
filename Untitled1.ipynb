## day-1 identifier-variable
### rules

```python
##use letters and digits and underscore (-)
name = "farhat"
age = 20
server_name = "aws"
_total = 500
print(name)
print(age)
print(server_name)
print(_total)
```

```python
##never start identifier with didgit
age=21
print(age)
```

```python
##do not use python keyword as identifiers
if_age = 10 
for_count = 20 
print(if_age)
print(for_count)
```

```python
##don not use space or hypens(-) in identifiers
server_name = "aws"
print(server_name)
```

```python
##prefer snake_case for names
server_name = "Web Server"
student_age = 20 
total_marks = 100
STUDENT_NAME = "Web Server"
print(server_name)
print(student_age)
print(total_marks)
print(STUDENT_NAME)
```

```python
##avoid using python built-in functions names as variables name 
print("hello")
my_number = 100
print(my_number)
```
